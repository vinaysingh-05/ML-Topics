````python


from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

data = load_breast_cancer()

X = data.data
y = data.target



X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)



scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

````

# Create SVM Model

````python
model = SVC(
    kernel="rbf",     # Default kernel
    C=1.0,            # Regularization
    gamma="scale"     # Default gamma
)

````

````python

model.fit(
    X_train,
    y_train
)



y_pred = model.predict(X_test)



print("Confusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nAccuracy")
print(accuracy_score(y_test, y_pred))

print("\nPrecision")
print(precision_score(y_test, y_pred))

print("\nRecall")
print(recall_score(y_test, y_pred))

print("\nF1 Score")
print(f1_score(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
````

# 📊 Normal SVM vs GridSearchCV SVM

| Feature | Normal SVM | GridSearchCV SVM |
|----------|------------|------------------|
| **Number of Models Trained** | Trains only one SVM model | Trains multiple SVM models with different parameter combinations |
| **Hyperparameters** | Uses manually selected or default parameters | Automatically searches for the best hyperparameters |
| **Training Time** | Faster | Slower because many models are evaluated |
| **Purpose** | Provides a baseline model | Optimizes the model for better performance |
| **Parameter Tuning** | Not performed | Performed automatically |
| **Cross Validation** | Not used by default | Uses Cross Validation (`cv`) to evaluate each parameter combination |
| **Best Parameters** | Not available | Available using `grid.best_params_` |
| **Best Model** | The manually created model | Available using `grid.best_estimator_` |
| **Example** | `SVC(kernel="rbf")` | `GridSearchCV(SVC(), param_grid=...)` |

---

## 📌 Conclusion

### ✅ Use **Normal SVM** when:
- You want a quick baseline model.
- You are learning SVM.
- You want fast training.

### ✅ Use **GridSearchCV** when:
- You want the best-performing model.
- You need to optimize hyperparameters (`kernel`, `C`, `gamma`, `degree`).
- You want a more reliable model using Cross Validation.
